# 玩具示例

## 问题

我们描述**背包问题的一个玩具示例**。我们考虑 8 件物品，其重量分别为 10、60、30、40、30、20、20、2，价值分别为 1、10、15、40、60、90、100、15。目标是确定一个物品子集，使其总重量小于 102，并且总价值尽可能大。

	

### 学到的建模原则

- 了解 Hexaly Optimizer 的建模风格：[区分决策变量与中间表达式](https://www.hexaly.com/docs/last/modelingprinciples/modelingprinciples.html#distinguish-decision-variables-from-intermediate-variables)


## 模型

我们将这个玩具问题建模为一个整数规划。对每件物品定义一个布尔决策变量，若该物品被选中则等于 1，否则等于 0。我们使用 **sum（求和）** 运算符计算所选物品的总重量。请注意，该值是由决策变量的取值推导出来的：总重量是一个中间表达式，而非决策变量。定义之后，我们可以约束总重量不超过 102。类似地，我们再定义一个对应所选物品总价值的中间表达式。最后，最大化该总价值。


## Python 实现


```python
# Copyright (c) Hexaly. Permission is hereby granted to use, copy,
# and modify this code for applications developed with Hexaly.
import hexaly.optimizer

with hexaly.optimizer.HexalyOptimizer() as optimizer:
    weights = [10, 60, 30, 40, 30, 20, 20, 2]
    values = [1, 10, 15, 40, 60, 90, 100, 15]
    knapsack_bound = 102

    #
    # Declare the optimization model
    #
    model = optimizer.model

    # 0-1 decisions
    x = [model.bool() for _ in range(8)]

    # Weight constraint
    knapsack_weight = model.sum(weights[i] * x[i] for i in range(8))
    model.constraint(knapsack_weight <= knapsack_bound)

    # Maximize value
    knapsack_value = model.sum(values[i] * x[i] for i in range(8))
    model.maximize(knapsack_value)

    model.close()

    # Parameterize the optimizer
    optimizer.param.time_limit = 10

    optimizer.solve()
```
