## 问题描述

**Branin 函数** 是一个二维函数,由如下方程定义:

$$f(x) = a\left(x_2 - b x_1^2 + c x_1 - r\right)^2 + s(1-t)\cos(x_1) + s$$

推荐的参数取值为 $a = 1$,$b = \frac{5.1}{4\pi^2}$,$c = \frac{5}{\pi}$,$r = 6$,$s = 10$,以及 $t = \frac{1}{8\pi}$。该函数通常在 $x_1 \in [-5, 10]$,$x_2 \in [0, 15]$ 的定义域上求值。

该函数具有三个全局最小值,问题的目标是找到其中之一。更多细节,请参阅 [Virtual Library of Simulation Experiments](http://www.sfu.ca/~ssurjano/branin.html)。

## 建模方法

常数取为推荐值:$a = 1$,$b = \frac{5.1}{4\pi^2}$,$c = \frac{5}{\pi}$,$r = 6$,$s = 10$,以及 $t = \frac{1}{8\pi}$。$x_1$ 与 $x_2$ 的定义域分别为 $[-5, 10]$ 与 $[0, 15]$。

Branin 函数问题的 Hexaly 模型非常直接。它使用两个浮点决策变量,分别表示 $x_1$ 与 $x_2$。借助 Hexaly 丰富的非线性数学算子集合,我们可以直接计算该函数的值。使用 **cos** 与 **pow** 等算子,我们能够以非常自然的方式表达 Branin 函数。
